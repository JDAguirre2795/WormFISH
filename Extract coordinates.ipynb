```python
import h5py
import numpy as np
```

```python
def extract_coordinates(h5_path, output_npy_path):
    """
    Extracts XYZ coordinates for each point in each frame from an HDF5 file,
    appends the frame index, and saves the result as a (T*N_points, 4) NumPy array.
    
    Parameters:
        h5_path (str): Path to the input HDF5 file.
        output_npy_path (str): Path to save the output .npy file.
        
    Returns:
        coords_with_frame (np.ndarray): Array of shape (T*N_points, 4)
                                        with [x, y, z, t] per row.
    """
    with h5py.File(h5_path, 'r') as f:
        points_dataset = f['points'][:]  # shape (T, N_points+1, 3)
        T = f.attrs['T']
        N_points = f.attrs['N_points']
        
        coords = points_dataset[:, :N_points, :]  # shape (T, N_points, 3)
        
        # Create an array for frame indices of shape (T, N_points, 1)
        frame_indices = np.arange(T).reshape(-1, 1, 1)
        frame_indices = np.tile(frame_indices, (1, N_points, 1))  # (T, N_points, 1)

        # Concatenate XYZ with frame index
        coords_with_frame = np.concatenate([coords, frame_indices], axis=2)  # (T, N_points, 4)
        
        # Flatten to (T*N_points, 4)
        coords_with_frame = coords_with_frame.reshape(-1, 4)

    np.save(output_npy_path, coords_with_frame)
    print(f"Saved [x, y, z, t] coordinates to {output_npy_path}")
    return coords_with_frame
```

```python
# Example usage
h5_path = R"C:\Users\josed\Documents\data\CompiledH5 MG2Pool1\pos1_remake.h5" # Replace with your H5 file path
output_npy_path = r"C:\Users\josed\Documents\data\POINTS-Coordinates\MG2Pool1\Pos1_points.npy"  # Output file name
extract_coordinates(h5_path, output_npy_path)
```

```python
# Load the saved .npy file
real =r"C:\Users\josed\Downloads\CompiledPos11_points.npy"
points = np.load(real)  # Shape: (T, N_points+1, 3)

# Print the full array
print("Full NumPy array:")
print(points)
```

```python
# Extract coordinates (X, Y, Z) and point IDs (N_points)
coordinates = points[:, :3]  # Columns 0-2: X, Y, Z
point_ids = points[:, 3]     # Column 3: Point IDs (N_points)

# Combine into a structured array for clarity
structured_data = np.column_stack((point_ids, coordinates))

# Print all entries (including NaN)
print("Point ID |    X       |    Y       |    Z")
print("-------------------------------------------")
for row in structured_data:
    print(f"{row[0]:7.1f} | {row[1]:9.2f} | {row[2]:9.2f} | {row[3]:5.1f}")
```
