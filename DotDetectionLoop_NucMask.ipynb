```python
import os
import numpy as np
import bigfish
import bigfish.stack as stack
import bigfish.detection as detection
import bigfish.multistack as multistack
import bigfish.plot as plot
print("Big-FISH version: {0}".format(bigfish.__version__))
from matplotlib import pyplot as plt
```

```python
from stardist.models import StarDist3D
from stardist.plot import render_label
from csbdeep.utils import normalize
```

```python
def process_tiff_file(input_path, output_path):
    # Load the image
    img = stack.read_image(input_path)
    
    DAPI = img[:,:,:,2]
    A647 = img[:,:,:,0]
    CY3B = img[:,:,:,1]
    
    # Detect spots and perform decomposition for A647 channel
    spotsA647, threshold = detection.detect_spots(
        images=A647,
        threshold= 15,
        return_threshold=True, 
        voxel_size=(1000,103 ,103),  # in nanometer (one value per dimension zyx)
        spot_radius=(1050, 300, 300))  # in nanometer (one value per dimension zyx)
    
    spots_A647_post, dense_regions, reference_spot = detection.decompose_dense(
        image=A647, 
        spots=spotsA647, 
        voxel_size=(1000, 103, 103), 
        spot_radius=(1050, 300, 300), 
        alpha=0.5,  # alpha impacts the number of spots per candidate region
        beta=1.7,  # beta impacts the number of candidate regions to decompose
        gamma=15)  # gamma the filtering step to denoise the image
    print("detected spots A647 before decomposition")
    print("\r shape: {0}".format(spotsA647.shape))
    print("\r dtype: {0}".format(spotsA647.dtype), "\n")
    print("detected spots A647 after decomposition")
    print("\r shape: {0}".format(spots_A647_post.shape))
    print("\r dtype: {0}".format(spots_A647_post.dtype))
    
    spotsCY3B, threshold = detection.detect_spots(
        images=CY3B,
        threshold = 25,
        return_threshold=True, 
        voxel_size=(1000,103 ,103),  # in nanometer (one value per dimension zyx)
        spot_radius=(1050, 300, 300))  # in nanometer (one value per dimension zyx)
    
    spots_CY3B_post, dense_regions, reference_spot = detection.decompose_dense(
        image=CY3B, 
        spots=spotsCY3B, 
        voxel_size=(1000, 103, 103), 
        spot_radius=(1050, 300, 300), 
        alpha=0.5,  # alpha impacts the number of spots per candidate region
        beta=1.7,  # beta impacts the number of candidate regions to decompose
        gamma=25)  # gamma the filtering step to denoise the image
    print("detected spots before decomposition CY3B")
    print("\r shape: {0}".format(spotsCY3B.shape))
    print("\r dtype: {0}".format(spotsCY3B.dtype), "\n")
    print("detected spots after decomposition CY3B")
    print("\r shape: {0}".format(spots_CY3B_post.shape))
    print("\r dtype: {0}".format(spots_CY3B_post.dtype))
    
    # Extract the base name of the file
    base_name = os.path.basename(input_path).replace(".tif", "")
    
    # Create subdirectory in the output path
    subdir = os.path.join(output_path, os.path.basename(os.path.dirname(input_path)))
    os.makedirs(subdir, exist_ok=True)
    
    # Save the results
    np.save(os.path.join(subdir, f"spots_A647_{base_name}.npy"), spots_A647_post)
    np.save(os.path.join(subdir, f"spots_CY3B_{base_name}.npy"), spots_CY3B_post)
    print(f"Processed and saved results for {base_name}")
```

```python
def process_directory(input_root, output_root):
    for subdir, _, files in os.walk(input_root):
        for file in files:
            if file.endswith(".tif"):
                input_path = os.path.join(subdir, file)
                try:
                    process_tiff_file(input_path, output_root)
                except:
                    print(f'failure on file {file}')
```

```python
def process_dapi_channel(input_path, output_path, model):
    # Load the image
    img = stack.read_image(input_path)
    
    # Extract the DAPI channel
    DAPI = img[:,:,:,2]
    print(DAPI.shape)
    
    # Determine the number of channels
    n_channel = 1 if DAPI.ndim == 3 else DAPI.shape[-1]
    axis_norm = (0,1,2)  # Normalize channels independently
    
    # Normalize the DAPI image
    img_normalized = normalize(DAPI, 1, 99.8, axis=axis_norm)
    
    # Predict instances using the Stardist model
    labels, details = model.predict_instances(img_normalized, n_tiles=(1, 12, 12))
    
    # Extract the base name of the file
    base_name = os.path.basename(input_path).replace(".tif", "")
    
    # Create subdirectory in the output path
    subdir = os.path.join(output_path, os.path.basename(os.path.dirname(input_path)))
    os.makedirs(subdir, exist_ok=True)
    
    # Save the labels
    np.save(os.path.join(subdir, f"Masks_{base_name}.npy"), labels)
    print(f"Processed and saved masks for {base_name}")
```

```python
def process_directory_DAPI(input_root, output_root):
    for subdir, _, files in os.walk(input_root):
        for file in files:
            if file.endswith(".tif"):
                input_path = os.path.join(subdir, file)
                process_dapi_channel(input_path, output_root, model)
```

```python
def process_directory_DAPI_hyb0(input_root, output_root, model):
    """
    Process only files starting with 'hyb_0' in all subfolders of input_root.
    Saves masks in subfolders under output_root corresponding to each position folder.
    """
    for subdir, _, files in os.walk(input_root):
        for file in files:
            if file.startswith("hyb_0") and file.endswith(".tif"):
                input_path = os.path.join(subdir, file)
                process_dapi_channel(input_path, output_root, model)
```
